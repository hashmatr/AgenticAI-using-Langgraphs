```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
```

```python
class QaState(TypedDict):
    Question:str
    Ans:str
```

```python
ChatModel = ChatOpenAI()

def ChatModelLLM(QaState) -> QaState:
    Ques = state["Question"]
    PROMPT = 'Answer the question {Ques}'
    answer = ChatModel.invoke(PROMPT).content
    state['Ans'] = answer

graph = StateGraph("Qastate")

graph.add_node('ChatModelLLM', ChatModelLLM)

graph.add_edge(START,ChatModel)
graph.add_edge(ChatModel,END)

workflow = graph.compile()

result = workflow.invoke({Question:'what is the capital of france'})

```

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict,Literal
```

```python
class calc_quadratic(TypedDict):
    a: int
    b: int
    c: int

    discriminant: float
    result: float
    equation: str 
```

```python
def show_equation(state: calc_quadratic):
    equation = f'{state["a"]}x2{state["b"]}x{state["c"]}'

    return {'equation': equation}

def calc_discriminant(state: calc_quadratic):
    discriminant = state["b"]**2 - 4 * state["a"] * state["c"]

    return {'discriminant': discriminant}

import math

def real_root(state: calc_quadratic):

    root1 = (-state["b"] + math.sqrt(state["discriminant"])) / (2 * state["a"])

    root2 = (-state["b"] - math.sqrt(state["discriminant"])) / (2 * state["a"])

    result = f'The roots are {root1} and {root2}'

    return {'result': result}

def not_real_root(state: calc_quadratic):


        result = f'No real root'
        return {'result': result}

def repeated_root(state: calc_quadratic):

        root1 = -(state["b"]/(2*state["a"]))

        result = f'The roots are {root1} and {root2}'
        return {'result': result}

def conditional(state: calc_quadratic) -> Literal["real_root","repeated_root","no_real_root"]:
    if state["discriminant"] > 0:
        return "real_root"
    elif state["discriminant"] == 0:
        return "repeated_root"
    else: 
        return "no_real_root"

```

```python
graph = StateGraph(calc_quadratic)
graph.add_node('show_equation', show_equation)
graph.add_node('calc_discriminant', calc_discriminant)
graph.add_node('real_root', real_root)
graph.add_node('not_real_root', not_real_root)
graph.add_node('repeated_root', repeated_root)

graph.add_edge(START, 'show_equation')
graph.add_edge('show_equation', 'calc_discriminant')

graph.add_conditional_edges(
	'calc_discriminant',
	conditional,
	{
		'real_root': 'real_root',
		'not_real_root': 'not_real_root',
		'repeated_root': 'repeated_root'
	}
)

graph.add_edge('real_root', END)
graph.add_edge('not_real_root', END)
graph.add_edge('repeated_root', END)
```

```python
workflow = graph.compile()

initial_state = {
    'a':2,
    'b':6,
    'c':3
}

workflow.invoke(initial_state)
```

```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
model = ChatOpenAI()
```

```python
load_dotenv()
```

```python
class BlogState(TypedDict):
    topic:str
    outline:str
    blog:str
```

```python
def outline(BlogState) ->BlogState:
    topic = State["topic"]
    prompt = "Create a perfect outline from this topic {topic}"
    outline = model.invoke(prompt).content
    state["outline"] = outline
    return outline
```

```python
def blog(BlogState) ->BlogState:
    outline = state["outline"]
    prompt = "Write a blog from this outline {outline}"
    blog = model.invoke(prompt).content
    state["blog"] = blog
    return blog
```

```python
graph = StateGraph(BlogState)

graph.add_node("outline", outline)
graph.add_node("blog", blog)

graph.add_edge(START,"outline")
graph.add_edge("outline","Blog")
graph.add_edge("Blog",END)

workflow = graph.compile()

result = workflow.invoke({'topic':"Pakistani politics"})
```
